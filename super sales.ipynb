# Datacleaning and Processing

```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["Ship Date"]=pd.to_datetime(df["Ship Date"])
df[["Order Date","Ship Date"]].dtypes
```

```python
text_columns=["Category","Sub-Category","Segment"]

for column in text_columns:
    df[column]=(
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
for column in text_columns:
    print("\n",column)
    print(df[column].unique)
```

```python
print("Final Shape:",df.shape)
print("\nTotal Missing Values:",df.isnull().sum().sum())
print("Total Duplicate Rows:",df.duplicated().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
numerical_features=[
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
]
df[numerical_features].describe()
```

```python
summary = pd.DataFrame({
    "Mean": df[numerical_features].mean(),
    "Median": df[numerical_features].median(),
    "Minimum": df[numerical_features].min(),
    "Maximum": df[numerical_features].max(),
    "Standard Deviation": df[numerical_features].std()
})

summary

```

# Category Wise Sales

```python
category_sales=(
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)
category_sales
```

# Category Wise Profit

```python
category_profit=(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
category_profit
```

# sub category wise profit

```python
sub_category_profit=(
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
sub_category_profit
```

# Region profit

```python
region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

region_profit
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()



    
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Sales",
)

plt.title("Sales Distribution")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Profit",
)

plt.title("Profit Distribution")
plt.ylabel("Total Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100
df[["Sales","Profit","Profit Margin"]].head()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit Margin",
)

plt.title("Discount vs Profit ")
plt.xlabel("Discount")
plt.ylabel("Profit Margin(%)")

plt.show()
```
